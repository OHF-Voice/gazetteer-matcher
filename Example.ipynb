```python
from gazetteer_matcher import GazetteerMatcher
from gazetteer_matcher.debug import render_interpretation

matcher = GazetteerMatcher()
```

```python
result = matcher.interpret("turn all the lights in the kitchen off, then set a 5 minute timer")
print(render_interpretation(result))
```

```python
result = matcher.interpret("turn all the lights in the kitchen off, then set a 5 minute timer")
result = matcher.interpret("turn them back on", previous_targets=result.targets)
print(render_interpretation(result))
```

```python
result = matcher.interpret("kill the lights")
print(render_interpretation(result))
```
