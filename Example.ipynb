```python
from gazetteer_matcher import GazetteerMatcher

matcher = GazetteerMatcher()
```

```python
result = matcher.interpret("turn all the lights in the kitchen off")
result.frames
```

```python
previous = matcher.interpret("open the bedroom blinds")
follow_up = matcher.interpret(
    "close them",
    previous_targets=previous.targets,
)
follow_up.frames
```

```python
rejected = matcher.interpret("start nothing")
rejected.rejection_code, rejected.response
```
